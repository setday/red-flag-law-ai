## 3. Explore Data

Let's examine some examples from the dataset.

```python
# Example legal clauses
examples = [
    "We reserve the right to terminate your account at any time without notice.",
    "You may cancel your subscription at any time through your account settings.",
    "The company is not liable for any damages arising from use of the service.",
    "We will notify you 30 days before any changes to these terms take effect."
]

for i, clause in enumerate(examples, 1):
    print(f"Example {i}: {clause}\n")
```

## 4. Analyze Clauses

Use the Red Flag Detector to analyze legal clauses.

```python
# Initialize detector
detector = RedFlagDetector()

# Analyze examples
for i, clause in enumerate(examples, 1):
    result = detector.analyze(clause)
    print(f"Example {i}:")
    print(f"  Text: {clause}")
    print(f"  Is Unfair: {result['is_unfair']}")
    print(f"  Category: {result['category']}")
    print(f"  Explanation: {result['explanation']}")
    print()
```
